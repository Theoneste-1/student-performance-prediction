```python
import pandas as pd
```

```python
data = pd.read_csv('dataset.csv')
data.head()
```

```python
def classify_student_category(hours_studied, previous_scores, extracurricular, sleep_hours, sample_papers):
    if(hours_studied < 0 or sleep_hours < 0 or previous_scores < 0 or sample_papers < 0):
        return "Invalid input data."
    elif(hours_studied + sleep_hours > 24):
        return "Impossible schedule."
    else:
        if(hours_studied == 0):
            return "No study = No marks (Strict mode)"
        else:
            return "Valid input data."
```

```python
data['category'] = data.apply(classify_student_category, axis=1)
```
